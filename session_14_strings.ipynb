- slice means cut into pieces

- it has similar analogy with for loop range

- in the for loop range we have start,stop,step

- range(start,stop,step) str[start:stop:step]

$case-1$

**str[:stop]**

- by default start=0

- by default step = positive direction 1

- last=stop-1

```python
str1='nikhil is honest'
print(str1[:6])
```

- str[start:]

- by default start=start value

- by default step = positive direction 1

- last=last index

- by default step = positive direction 1

- last=stop-1

**sliding window approach**

- fix the slice value first: **new[i:i+1]**

- then increase the slice by 1 : **for i in range(len(new))**

- whenerver it is equal to substring : **new[i:i+3]=='ola'**

- then update thee count: **count=count+1**

```python
new='ola ola ola'
count=0
for i in range(len(new)):
    if new[i:i+3]=='ola':
        count+=1
print(count)
```
