**find second max and second min**

```python
list5=[10,20,99,3,105,23,5,75]
maxx=list5[0]
second_max=list5[0]
for i in range(len(list5)):
    if list5[i]>maxx:
        maxx=list5[i]

for i in range(len(list5)):
    if list5[i]!=maxx:
        if list5[i]>second_max:
            second_max=list5[i]
print(second_max)

    
        
```

```python
list5=[10,20,99,3,105,23,5,75]
minn=list5[0]
second_min=list5[0]
for i in range(len(list5)):
    if minn>list5[i]:
        minn=list5[i]
print(f'First min number is {minn}')
for i in range(len(list5)):
    if list5[i]!=minn:
        if second_min>list5[i]:
            second_min=list5[i]
print(f'second min number is {second_min}')
```

```python
fname=[]
sname=[]
cname=[]
l='virat.kohali@rcb.com;rohit.sharma@mi.com;ms.dhoni@csk.com'
emails=l.split(';')
for i in emails:
    name,company=i.split('@')
    first,second=name.split('.')
    company=company.split('.')[0]
    fname.append(first)
    sname.append(second)
    cname.append(company)
print(fname)
print(sname)
print(cname)
```

```python
l=[]
unique_word=[]
l1='can canner can you can not canner with you but you can the canner'
word=l1.split()
for i in word:
    if i not in unique_word:
        unique_word.append(i)
for i in unique_word:
    l.append(f'{i}-{word.count(i)}')
print(l)
```
