```python
import random
i=0
counte=0
counto=0
sume=0
sumo=0
while i<5:
    n=random.randint(1,100)
    if n%2==0:
        print(f"{n} is even number")
        sume+=n
        counte+=1
    else:
        print(f"{n} is odd number")
        sumo+=n
        counto+=1
    i+=1

print(f"{counte} even numbers")
print(f"{counto} odd numbers")
print(f"{sume} even numbers")
print(f"{sumo} odd numbers")
```
